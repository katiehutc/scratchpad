```python
import pandas as pd

df = pd.read_csv('/content/data/craiglist_cville_cars_long.csv', encoding = 'latin1')
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

tdf = df.loc[:, ['price', 'age', 'miles']]
print(tdf.shape)
tdf = tdf.dropna()
print(tdf.shape)

y = tdf['price']
x = tdf.drop('price', axis=1)

model = model.fit(x, y)
y_hat = model.predict(x)
```
